## Data Aggregation

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
'data': range(6)}, columns=['key', 'data'])
print(df)
```

```python
df.groupby("key").sum()
```

```python
df1 = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
'data': range(6),"data1":[10,20,40,50,70,100]})
```

```python
df1
```

```python
df1.groupby("key").sum()
```

```python
df_sum=df1.groupby("key").aggregate({"data":"sum","data1":"max"})
```

```python
df_sum
```

## Example

```python
df=yf.download("^HSI","2020-01-01")
```

```python
df.tail()
```

```python
df1=pd.DataFrame({"High":df["High"]["^HSI"],"Low":df["Low"]["^HSI"]})
```

```python
df1
```

```python
df1["Date1"]=df1.index
```

```python
df1["Year"]=df1["Date1"].dt.year
df1["Month"]=df1["Date1"].dt.month
```

```python
df1
```

```python
dfM=df1.groupby(["Year","Month"]).aggregate({"High":"max","Low":"min"})
```

```python
dfM
```

```python
dfM.reset_index(inplace=True)
```

```python
dfM.head()
```

```python
dfM["Range"]=dfM["High"]-dfM["Low"]
```

```python
dfM
```

```python
dfM.groupby("Month").aggregate({"Range":"mean"})
```

```python
dfM.groupby("Month").aggregate({"Range":["mean","max","min","std"]})
```
