## Use of find and findAll() method

```python
html = requests.get("http://www.pythonscraping.com/pages/warandpeace.html")
bsObj = BeautifulSoup(html.text, "html.parser")
nameList = bsObj.findAll("span", {"class":"green"})
```

```python
nameList
```

```python
L=[]
for i in nameList:
    L.append(i.get_text().replace("\n"," "))
```

```python
L
```

```python
L=[i.get_text().replace("\n"," ") for i in nameList]
L    
```

## Use of selenium

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

#driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
```

```python
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
```

```python
driver.get("https://hk.finance.yahoo.com/gainers")
```

```python
html=driver.page_source
```

### Get HTML table

```python
bsObj = BeautifulSoup(html, "html.parser")
table1=bsObj.find("table", {"class":"yf-1bwepqw bd"})
```

```python
type(table1)
```

```python
df=pd.read_html(str(table1),header=0)[0]
```

```python
df.head()
```

```python
soup = BeautifulSoup(html, "lxml")
sz = soup.find("div", {"class":"rightpane itemtrad"})
szsb = sz.find("div", {"class":"row southbound"})
szsbval = szsb.find("div", {"class":"type value"})
val_str = szsbval.get_text()
val = int(val_str.replace("HK$", "").replace(",", "").replace("M", ""))*1000000
val
```
